```python
# Archivo
archivo = "sample_data/1.xlsx"

# Leer archivo completo
ciudades_df = pd.read_excel(archivo, skiprows=7, nrows=10)
# Reemplazar valores nulos por 0 (puedes cambiar por "Sin dato" u otro valor)
ciudades_df = ciudades_df.fillna(0)
ciudades_df = ciudades_df.rename(columns={"Unnamed: 0": "Num","Unnamed: 1": "Ciudad"})
ciudades_df

```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Supongamos que tu DataFrame ya se llama df
# Nos quedamos solo con las columnas necesarias
df_cascada = ciudades_df[["Ciudad", "Variación.2", "Contribución.5"]]

# Orden para que el Total nacional esté primero
df_cascada["Ciudad"] = pd.Categorical(df_cascada["Ciudad"],
                                      categories=["Total nacional"] +
                                      [c for c in df_cascada["Ciudad"] if c != "Total nacional"],
                                      ordered=True)
df_cascada = df_cascada.sort_values("Ciudad")

# Crear gráfico de cascada
fig, ax1 = plt.subplots(figsize=(10, 6))

# Cálculo de posiciones
cumulative = df_cascada["Variación.2"].cumsum() - df_cascada["Variación.2"]

# Colores: verde si positivo, rojo si negativo
colors = ['green' if val >= 0 else 'red' for val in df_cascada["Variación.2"]]

# Barras de variación
ax1.bar(df_cascada["Ciudad"], df_cascada["Variación.2"],
        bottom=cumulative, color=colors, label="Variación")

ax1.set_ylabel("Variación (%)", color="black")
ax1.set_xlabel("Ciudad")
ax1.tick_params(axis="x", rotation=45)

# Segundo eje para contribución
ax2 = ax1.twinx()
ax2.plot(df_cascada["Ciudad"], df_cascada["Contribución.5"],
         color="blue", marker="o", label="Contribución")
ax2.set_ylabel("Contribución (%)", color="blue")
ax2.yaxis.set_major_formatter(FuncFormatter(lambda y, _: f'{y:.1f}%'))

# Leyendas
ax1.legend(loc="upper left")
ax2.legend(loc="upper right")

plt.title("Gráfico de Cascada: Variación y Contribución")
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
df8 = pd.read_excel('/content/DANE_comercio electrónico - comercio al por menor - total nacional.xlsx')
df8.head(20)
```

```python
# Conocer cuántos datos null hay en la base de datos
df8.isnull().sum()

#Rellenar las casillas que estén vacías (null)
df8.fillna(method='ffill', inplace=True)
df8.isnull().sum()

# Eliminar duplicados
df8.drop_duplicates(inplace=True)

# Convertir nombres de meses en español a números
meses_es = {
    'Enero': 1, 'Febrero': 2, 'Marzo': 3, 'Abril': 4,
    'Mayo': 5, 'Junio': 6, 'Julio': 7, 'Agosto': 8,
    'Septiembre': 9, 'Octubre': 10, 'Noviembre': 11, 'Diciembre': 12
}

# Convertir a número de mes
df8['Mes'] = df8['Mes'].str.strip()
df8['Mes_num'] = df8['Mes'].map(meses_es)

# Crear fecha usando año y mes
df8['Fecha'] = pd.to_datetime(dict(year=df8['Año'].astype(int), month=df8['Mes_num'], day=1))
df8 = df8.sort_values('Fecha')

df8[['Año', 'Mes', 'Mes_num', 'Fecha']].head()

# Filtrar solo 2025
df8_2025 = df8[df8['Año'] == 2025]
df8.head(20)
```

```python

# Configuración
plt.figure(figsize=(10,6))
bar_width = 0.25
x = range(len(df8_2025))

# Barras
plt.bar([i - bar_width for i in x],
        df8_2025["Ventas realizadas a través de comercio electrónico"],
        width=bar_width, label="Ventas realizadas a través de comercio electrónico")

plt.bar(x,
        df8_2025["Ventas realizadas a través de canales diferentes al comercio electrónico"],
        width=bar_width, label="Ventas realizadas a través de canales diferentes al comercio electrónico")

# Personalización
plt.xticks(x, df8_2025["Mes"], rotation=45, ha="right")
plt.ylabel("Ventas totales")
plt.title("Comportamiento de las ventas digitales VS no digitales - 2025")
plt.legend()
plt.tight_layout()

# Mostrar gráfica
plt.show()
```
